```python
from unsloth import FastLanguageModel
import torch

model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = "unsloth/Ministral-3-8B-Instruct-2512",
    max_seq_length = 8192,
    dtype = None,          # auto-detect
    load_in_4bit = True,
)
FastLanguageModel.for_inference(model)  # enables faster inference mode
```

#SINHALA LANGUAGE

```python
import json
import re
import torch

# ── Load data ────────────────────────────────────────────────
DATA_PATH = "/content/sample_data/sri_lankan_simplified.json"
with open(DATA_PATH, "r", encoding="utf-8") as f:
    data = json.load(f)
print(f"Loaded {len(data)} items")

# ── Prompt building ──────────────────────────────────────────
INSTRUCTION_PROMPT = """You are a Sinhala Expert for answering multiple-choice questions.
You need to always evaluate these questions based on the Sri Lankan context and provide the most accurate answer.
You should only respond with the correct option letter, without any additional explanation or commentary.
Option C: පිළිතුරු දෙකම නිවැරදියි.
Option D: පිළිතුරු දෙකම නිවැරදි නොවේ.
Return only the correct option letter: 'A', 'B', 'C', or 'D'.
"""

def build_prompt(item):
    lines = [INSTRUCTION_PROMPT, "", item["scenario_question"], "", "Options:"]
    for letter, text in item["options"].items():
        lines.append(f"{letter}. {text}")
    return "\n".join(lines)

# ── Generation ────────────────────────────────────────────────
def generate_answer(prompt, item, max_new_tokens=20):
    messages = [{"role": "user", "content": prompt}]

    text = tokenizer.apply_chat_template(
        messages, add_generation_prompt=True, tokenize=False,
    )
    inputs = tokenizer(text=text, return_tensors="pt").to(model.device)

    with torch.no_grad():
        output = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,   # deterministic — reproducible eval results
        )

    input_len = inputs["input_ids"].shape[-1]
    raw = tokenizer.decode(output[0][input_len:], skip_special_tokens=True).strip()

    # C and D are fixed options present in every prompt, not stored in item["options"]
    valid_letters = list(item["options"].keys()) + ["C", "D"]
    predicted_letter = None

    match = re.search(r"\b([A-D])\b", raw.upper())
    if match and match.group(1) in valid_letters:
        predicted_letter = match.group(1)
    else:
        for letter, opt_text in item["options"].items():
            if opt_text.strip() and opt_text.strip() in raw:
                predicted_letter = letter
                break

    if predicted_letter is None:
        print(f"⚠️ No match for id={item.get('id')}: raw output was: {raw!r}")
        return None, raw

    if predicted_letter == "C":
        final_answer = "Both"
    elif predicted_letter == "D":
        final_answer = "0"
    else:
        final_answer = predicted_letter

    return final_answer, raw

# ── Main loop (resume-safe) ──────────────────────────────────
OUTPUT_PATH = "/content/sri_lankan_results.json"

for i, item in enumerate(data):
    if item.get("model_answer"):
        continue

    prompt = build_prompt(item)
    predicted, raw_output = generate_answer(prompt, item)

    item["model_answer"] = predicted
    item["raw_model_output"] = raw_output

    with open(OUTPUT_PATH, "w", encoding="utf-8") as f:
        json.dump(data, f, ensure_ascii=False, indent=2)

    print(f"[{i+1}/{len(data)}] {item['id']} -> {predicted} (gold: {item['gold_answer']})")

print("All done.")
```

```python
import json

with open("/content/sri_lankan_results.json", "r", encoding="utf-8") as f:
    results = json.load(f)

correct = sum(1 for item in results if item.get("model_answer") == item.get("gold_answer"))
total = len(results)

print(f"Correct: {correct} / {total}")
print(f"Accuracy: {correct / total:.2%}")
```
